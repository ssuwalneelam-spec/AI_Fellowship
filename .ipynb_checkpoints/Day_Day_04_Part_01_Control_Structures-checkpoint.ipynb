### Write a program to find the factorial of number using for loop

```python
number= int(input("Enter a number"))
fact=1
for i in range (1, number+1):
    fact = fact * i
print(f"{number}!={fact}")
```

```python
number= int(input("Enter a number"))
fact = 1
i = 1

while i <= number:
    fact = fact * i
    i= i+1
print(f"{number}! = {fact}")
```
