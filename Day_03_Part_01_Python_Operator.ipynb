Operators in Python

Operator: a special symbol or key keyword that performs an operation. Operand: the Value (or variable) the operator work on.

Classification by Number of Operands

1. Unary: One Operand (-,+,~,not)
2. Binary: Two operands (*, ==, and)
3. Ternary: three operands (the conditional expression x if condition else y)
