## LAMBDA FUNCTIONS (one-line operations)

```python
add = lambda a, b: a + b
print(add(1,2))
```

```python
sub = lambda a, b: a - b
print (sub(2,1))
```

```python
mul = lambda a, b: a * b
print(mul(1,2))
```

```python
square = lambda x: x ** 2
print(square(2))
```

```python
cube = lambda x: x ** 3
print(cube(2))
```
